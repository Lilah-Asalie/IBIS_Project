Datetime and latitude/longitude NaN values

```python
# NEW DATA

#convert to datetime
df['DATE'] = pd.to_datetime(df['DATE'],format = 'mixed',utc=True)
df = df.dropna(subset=["DATE"]) #drop any cycles with no date or lat/lon 

#fix lat lon values with '--' (replace with nan) 
df['LATITUDE'] = pd.to_numeric(df['LATITUDE'], errors='coerce')
df['LONGITUDE'] = pd.to_numeric(df['LONGITUDE'], errors='coerce')
```

Month, year, and season column added

```python
#create season columns
df["year"] = df["DATE"].dt.year.astype("Int64")
df["month"] = df["DATE"].dt.month.astype("Int64")

def assign_season(row):
    m = int(row["month"])
    y = int(row["year"])
    if m >= 9:
        return f"{y}-{y+1}"
    elif m <= 4:
        return f"{y-1}-{y}"
    else:
        return f"{y}"

df["season"] = df.apply(assign_season, axis=1)
```

Physical variable columns added (SA, CT, potential density, in-situ density, buoyancy frequency)

```python
# define function for calculating SA, CT, potential dens, in-situ dens, buoyancy frequency
def calc_phys_vars(df):
        
    # Columns needed
    cols = ['PSAL_ADJUSTED','TEMP_ADJUSTED' ,'PRES_ADJUSTED', 'LATITUDE', 'LONGITUDE']
    
    # only do calculation on good qc T and S data - create nan values where not good data
    qc_flags = [1,2,5,8]
    mask_qc = df['PSAL_ADJUSTED_QC'].isin(qc_flags) & df['TEMP_ADJUSTED_QC'].isin(qc_flags)
    df['PSAL_ADJUSTED'] = df['PSAL_ADJUSTED'].where(mask_qc)
    df['TEMP_ADJUSTED'] = df['TEMP_ADJUSTED'].where(mask_qc)    
    
    # mask valid (non-null) rows
    mask = df[cols].notnull().all(axis=1)
    
    # initialize outputs as NaN
    SA = np.full(len(df), np.nan)
    CT = np.full(len(df), np.nan)
    rho = np.full(len(df), np.nan)
    sigma0 = np.full(len(df), np.nan)
    nsq = np.full(len(df), np.nan)
    
    # extract numeric arrays for valid rows
    SP = df.loc[mask, 'PSAL_ADJUSTED'].astype(float).to_numpy()
    PT = df.loc[mask, 'TEMP_ADJUSTED'].astype(float).to_numpy()
    pres = df.loc[mask, 'PRES_ADJUSTED'].astype(float).to_numpy()
    LAT = df.loc[mask, 'LATITUDE'].astype(float).to_numpy()
    LON = df.loc[mask, 'LONGITUDE'].astype(float).to_numpy()
    
    # calculate variables using gsw
    SA_valid = gsw.SA_from_SP(SP=SP, p=pres, lon=LON, lat=LAT)
    CT_valid = gsw.CT_from_t(SA=SA_valid, t=PT, p=pres)
    rho_valid = gsw.rho(SA=SA_valid, CT=CT_valid, p=pres)
    sigma0_valid = gsw.sigma0(SA=SA_valid, CT=CT_valid)
    nsq_valid = gsw.Nsquared(SA=SA_valid, CT=CT_valid, p=pres,lat=LAT)
    
    # store back into full-length arrays
    SA[mask] = SA_valid
    CT[mask] = CT_valid
    rho[mask] = rho_valid
    sigma0[mask] = sigma0_valid
    np.putmask(nsq, mask, nsq_valid) # found on stack overflow because nsq was 2 dimensional so previous method wasn't working
    
    # sdd results to the dataframe
    df = df.copy()
    df['SA'] = SA
    df['CT'] = CT
    df['rho'] = rho
    df['sigma0'] = sigma0
    df['nsq'] = nsq
    
    return df

df_edited = calc_phys_vars(df)
```

MLP column added - eventually change how it's calculated?

```python
def calc_mlp_density(df_profile, delta_sigma=0.03):
    #calc MLD for a single profile using density threshold
    df_profile = df_profile.sort_values('PRES_ADJUSTED')

    # get SA, CT, and potential density values
    SA = df_profile['SA']
    CT = df_profile['CT']
    sigma0 = df_profile['sigma0']

    sigma_surface = sigma0.iloc[0]
    d_sigma = sigma0 - sigma_surface

    # find first depth where density difference exceeds threshold
    idx = np.argmax(d_sigma > delta_sigma)
    if d_sigma.iloc[idx] <= delta_sigma:
        return np.nan  # no threshold reached
    return df_profile['PRES_ADJUSTED'].iloc[idx]

mld = (df_edited.groupby(['WMOID', 'CYCLE_NUMBER']).apply(calc_mlp_density).reset_index(name='MLP'))
df_full_edited = df_edited.merge(mld, on=['WMOID', 'CYCLE_NUMBER'], how='left')
```

```python
def parse_argo_csv(filepath):
    float_id = None
    profile_number = None
    profile_date = None
    header_line_idx = None

    with open(filepath, "r") as f:
        lines = f.readlines()

    for i, line in enumerate(lines): # add elif lines for lat/lon
        if line.startswith("#platform_number"):
            float_id = line.split("=")[1].strip()
        elif line.startswith("#cycle_number"):
            profile_number = int(line.split("=")[1].strip())
        elif line.startswith("#profile_date"):
            profile_date = line.split("=")[1].strip()
        elif not line.startswith("#"):
            header_line_idx = i
            break

    df = pd.read_csv(filepath, skiprows=header_line_idx)

    # Strip units in parentheses from column names
    df.columns = [re.sub(r"\s*\(.*?\)", "", col).strip() for col in df.columns]

    df["float_id"] = float_id
    df["profile_number"] = profile_number
    df["profile_date"] = pd.to_datetime(profile_date)
    # add columns for lat/lon

    return df

# Example: one float
# df = parse_argo_csv(filepath = '/Users/steviewalker/Documents/Argo/EasyOneArgoBGCLite/EasyOneArgoBGCLite_20260914T035253Z/data/5905062/5905062_001_EasyBGCLite.csv')
df = parse_argo_csv(filepath = '/Users/lilah/Documents/IBIS_Project/Data_Summer/EasyOneArgoBGCLite_20260914T035253Z/data/5905103/5905103_001_EasyBGCLite.csv')

# print(df.columns.tolist())
```

```python
import glob
import os

def combine_float_profiles(folder_path, output_path=None):
    csv_files = sorted(glob.glob(os.path.join(folder_path, "*_EasyBGCLite.csv")))

    if not csv_files:
        raise FileNotFoundError(f"No matching CSV files found in {folder_path}")

    dfs = [parse_argo_csv(f) for f in csv_files]
    combined = pd.concat(dfs, ignore_index=True)

    # Sort by profile number so profiles are in order
    combined = combined.sort_values("profile_number").reset_index(drop=True)

    if output_path:
        combined.to_csv(output_path, index=False)

    return combined

# Example: combine all profiles for float 5905062
# folder = '/Users/steviewalker/Documents/Argo/EasyOneArgoBGCLite/EasyOneArgoBGCLite_20260914T035253Z/data/5905104'
folder = '/Users/lilah/Documents/IBIS_Project/Data_Summer/EasyOneArgoBGCLite_20260914T035253Z/data/5905103'
combined = combine_float_profiles(folder, output_path=os.path.join(folder, "5905103_combined.csv"))

print(combined.shape)
print(combined["profile_number"].unique())
```
